```python
import pandas as pd

# 1. Load Data
df = pd.read_excel(r"C:\Users\User\Downloads\Sample data (1).xlsx")

# 2. Data Cleaning
df.columns = df.columns.str.strip()  # Fix column header spaces
df['Discount Band'] = df['Discount Band'].fillna('None') # Clean missing discount band values

# 3. Overall Financial KPIs
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_cogs = df['COGS'].sum()
total_discounts = df['Discounts'].sum()
total_units = df['Units Sold'].sum()
profit_margin = (total_profit / total_sales) * 100

print(f"=== OVERALL KPIS ===")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total COGS: ${total_cogs:,.2f}")
print(f"Total Discounts: ${total_discounts:,.2f}")
print(f"Total Units Sold: {total_units:,.1f}")
print(f"Overall Profit Margin: {profit_margin:.2f}%\n")

# 4. Aggregations for Insights
print("=== PERFORMANCE BY SEGMENT ===")
segment_df = df.groupby('Segment').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Units_Sold=('Units Sold', 'sum')
).reset_index()
segment_df['Profit_Margin_%'] = (segment_df['Profit'] / segment_df['Sales']) * 100
print(segment_df.sort_values(by='Sales', ascending=False).to_string(index=False))

print("\n=== PERFORMANCE BY PRODUCT ===")
product_df = df.groupby('Product').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Units_Sold=('Units Sold', 'sum')
).reset_index()
print(product_df.sort_values(by='Sales', ascending=False).to_string(index=False))

print("\n=== PERFORMANCE BY COUNTRY ===")
country_df = df.groupby('Country').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
).reset_index()
print(country_df.sort_values(by='Sales', ascending=False).to_string(index=False))
```
